# Literature triangulation

Literature có thật sự xác nhận hiện tượng đó không ? Paper nào xác nhận trực tiếp ? Paper nào chỉ hỗ trợ gián tiếp ? Paper nào không kiểm chứng nó ? Có evidence nào khiến ta phải thận trọng ?

4 Paper:

| Mã | Paper |
|---|---|
| **P1 — ViHOS** | Hoang et al. (2023), **“ViHOS: Hate Speech Spans Detection for Vietnamese”**, EACL 2023 |
| **P2 — ViSoBERT** | Nguyen et al. (2023), **“ViSoBERT: A Pre-Trained Language Model for Vietnamese Social Media Text Processing”**, EMNLP 2023 |
| **P3 — VLUE/CafeBERT** | Do et al. (2024), **“VLUE: A New Benchmark and Multi-task Knowledge Transfer Learning for Vietnamese Natural Language Understanding”**, Findings of NAACL 2024 |
| **P4 — ViHateOff** | Vu & Le (2026), **“Enhanced framework for detecting Vietnamese hate and offensive spans”**, IJ-AI 2026 |

P1 là paper gốc của ViHOS; P2 nghiên cứu PLM cho Vietnamese social media; P3 đưa ViHOS vòa benchmark BLUE và đề xuất CafeBERT; P4 trực tiếp xây ViHateOff trên ViHOS bằng PhoBERT-Large + hated-speech dictionary

# 1. O1: Offensive meaning phụ thuộc context

Evidence của chính nhóm: M5.1 đã tìm ra các case như S05, S13, S18, S20, S32, S44, S45. Ví dụ người khuyết tật, nó, mày, thằng, bọn không mang nhãn offensive cố định khi đứng riêng; vai tro của chúng thay đổi theo toàn phát ngôn

Đối chiếu literature:

| Paper | Evidence | Đánh giá |
|---|---|---|
| **P1 — ViHOS** | Error analysis có hẳn category **“lack of context”**; paper nói wrong span prediction, wrong boundary, allusion, annotation error và lack of context là các loại lỗi lớn của PhoBERT-Large và XLM-R-Large | **TRỰC TIẾP** |
| **P2 — ViSoBERT** | Dùng contextualized representations và đánh giá trên hate-speech span detection, nhưng không tách riêng các case “context-dependent insult” | **GIÁN TIẾP** |
| **P3 — VLUE/CafeBERT** | CafeBERT/XLM-R/PhoBERT đều là contextual PLM và được đánh giá trên ViHOS, nhưng paper không có subgroup “context-dependent” | **GIÁN TIẾP yếu** |
| **P4 — ViHateOff** | Kiến trúc vẫn giữ PhoBERT-Large để biểu diễn quan hệ/context giữa token rồi bổ sung lexical dictionary; nhưng không chạy riêng experiment trên context-dependent cases | **GIÁN TIẾP** |

**P1 — ViHOS là evidence mạnh nhất ở đây**
- P1 - ViHOS không chỉ nói rằng contexxt quan trọng. Trong error analysis trên dev, hai model mạnh PhoBERT-Large và XLM-R-Large gặp các lỗi lớn bao gồm lack of context và allusion. P1 định nghĩa `lack of context` là trường hợp comment có thể được hiểu theo nhiều nghĩa ở các context khác nhau

**P2 — ViSoBERT cho gì?**
- P2 - ViSoBERT cung cấp contextual embeddings từ một PLM pretrain trên social-media text và còn dùng chúng làm input feature cho BiLSTM/BiGRU. Paper cho thấy embeddings của ViSoBERT tốt hơn các pretrained embeddings so sánh trên các downstream social-media tasks

Kết luận literature cho O1:
- O1 được literature hỗ trợ mạnh. Đặc biệt P1 - ViHOS xác nhận trực tiếp rằng thiếu context là một nguồn lỗi của span detection. P2 - ViSoBERT, P3 - VLUE/CafeBERT và P4-ViHateOff cho thấy context PLM là thành phần hợp lý, nhưng chưa paper nào trong ba paper đó isolate riêng contextual-insult cases

# 2. O2: Figurative/metaphorical insults là một difficulty riêng

M5.1 có evidence rất rõ: S09, S15, S26, S27, S35, S40, S42, S47, S49.

Ví dụ:
- não bị vi rút ăn hết
- lắp não vào

Không thể hiểu chúng bằng literal word meaning đơn giản:

| Paper | Evidence | Đánh giá |
|---|---|---|
| **P1 — ViHOS** | Error taxonomy có `allusion`, `idiom/proverb`, `metaphor/metonymy`; allusion còn nằm trong nhóm major errors | **TRỰC TIẾP** |
| **P2 — ViSoBERT** | Không có targeted metaphor/figurative analysis | **CHƯA CÓ** |
| **P3 — VLUE/CafeBERT** | Không có targeted metaphor analysis | **CHƯA CÓ** |
| **P4 — ViHateOff** | Introduction nhắc irony/slang/sarcasm là thách thức tiếng Việt, nhưng không đánh giá ViHateOff riêng trên figurative cases | **GIÁN TIẾP rất yếu** |

**P1 - ViHOS xác nhận rất rõ phenomenon này**
- Error taxonomy của P1 - ViHOS có riêng:
    - allusion
    - idiom/proverb
    - metaphor/metonymy
    - lack of context
- P1 còn ghi allusion trong các nhóm prediction failure chính của hai model mạnh. 

# 3. O3: Social-media noise la một property quan trọng

M5.1 ta có:
- S01 ko
- S02 punctuation kiểu ----CÚT
- S07 mắc dạy

| Paper | Evidence | Đánh giá |
|---|---|---|
| **P1 — ViHOS** | Error categories gồm abbreviation, spelling mistake, missing diacritics, rare/mixed/stuck words | **TRỰC TIẾP** |
| **P2 — ViSoBERT** | Pretrained riêng trên Vietnamese social media; thực nghiệm riêng về teencode, emoji, missing diacritics | **TRỰC TIẾP RẤT MẠNH** |
| **P3 — VLUE/CafeBERT** | Vietnamese continued pretraining cải thiện ViHOS, nhưng corpus là Wikipedia + news, không phải social media | **ĐỐI CHIẾU QUAN TRỌNG** |
| **P4 — ViHateOff** | HSD có thể ghi nhớ các slang/distorted forms đã xuất hiện trong training; paper claim lợi ích nhưng chưa ablation | **GIÁN TIẾP** |

**P2 — ViSoBERT là literature evidence mạnh nhất cho O3**
- Đây là điểm phải hiểu thật rõ vì sau này nếu ta thử ViSoBERT thì lý do nằm ở đây, chứ không phải vì tên nó nổi tiếng.
- P2 - ViSoBERT lập luận rằng general-domain Vietnamese PLMs không hoàn toàn đại diện các sociolinguistic phenomena của social media. Paper liệt kê informal language, non-standard vocabulary, missing diacritics, emoji, teencode

**P3 — VLUE/CafeBERT tạo ra một control cực kỳ quan trọng**

- Nếu ta chỉ đọc P2, ta rất dễ suy nghĩ: ViHOS là social media → social-media PLM chắc chắn tốt nhất.
- Nhưng P3 - VLUE/CafeBERT khiến ta phải thận trọng: CafeBERT không được continued-pretrain trên Vietnamese social media. Corpus mà P3-VLUE/CafeBERT dùng là:
    - 1GB Vietnamese Wikipedia;
    - 17GB Vietnamese online news.
- Tức là P3 là Vietnamese-adapted nhưng không social-media-specific.

# 5. O5: Exact span boundary là difficulty độc lập

| Paper | Evidence | Đánh giá |
|---|---|---|
| **P1 — ViHOS** | `wrong spans boundary` là một trong các major prediction errors; annotation guideline nhấn mạnh span extent | **TRỰC TIẾP RẤT MẠNH** |
| **P2 — ViSoBERT** | Preprocessing phải cập nhật lại span positions để không phá alignment | **GIÁN TIẾP** |
| **P3 — VLUE/CafeBERT** | Chỉ report ViHOS task F1, không boundary-error breakdown | **CHƯA CÓ** |
| **P4 — ViHateOff** | Report span performance nhưng không phân tích riêng boundary error | **CHƯA CÓ** |

**P1 — ViHOS gần như xác nhận chính xác thứ ta thấy**

- P1 định nghĩa: wrong spans boundary = model dự đoán boundary không đầy đủ hoặc sai.
- Thậm chí annotation guideline của P1 có những ví dụ nhấn mạnh không được gom hoặc cắt span tùy ý, tức chính annotation task đã coi extent của span là một phần quan trọng

# 6. O6: Clean examples là control quan trọng chống surface shortcut

| Paper | Evidence | Đánh giá |
|---|---|---|
| **P1 — ViHOS** | Thêm 5,528 clean comments làm performance tăng đáng kể; authors nói giảm bias | **TRỰC TIẾP RẤT MẠNH** |
| **P2 — ViSoBERT** | Không có clean-control ablation | **CHƯA CÓ** |
| **P3 — VLUE/CafeBERT** | Không có clean-control analysis | **CHƯA CÓ** |
| **P4 — ViHateOff** | Không có clean-vs-lexical false-positive ablation | **CHƯA CÓ** |

P1 - ViHOS so: training data without additional clean comments và Full Data + 5,528 additional clean comments. Paper báo performance cải thiện trung bình khoảng 0.1002 ± 0.0210 F1; riêng XLM-R-Large tăng 0.1014 và PhoBERT-Large tăng 0.0849. Authors kết luận additional clean comments giúp giảm model bias và cải thiện performance.

O6 được P1 — ViHOS xác nhận trực tiếp bằng experiment. Clean controls không chỉ có giá trị minh họa; chúng có giá trị trong việc giảm bias của model.
