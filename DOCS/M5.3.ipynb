# Chốt các giả thuyết có thể kiểm chứng dựa trên M5.1 + M5.2

# 1. H1 - Contextual Modeling Hypothesis

**Ta bắt đầu từ evidence**
- Từ M5.1:
    - S05: có vấn đề về quang học chỉ mang sắc thái mỉa mai khi đọc cả context.
    - S13: người khuyết tật bản thân là cụm trung tính nhưng được dùng để hạ thấp trong câu.
    - S18, S20, S32, S44, S45: các từ như thằng, mày, con, nó, bọn không có giá trị offensive cố định ngoài context.
    - Ngược lại các clean cases S03, S25, S37, S39, S48 cho thấy negative/violent/hostile-looking words chưa chắc offensive.
- Do đó ta đang thấy hai chiều của cùng một vấn đề:
    - Token trông bình thường + context => có thể offensive
    - Token trông tiêu cực + context => vẫn có thể clean

Literature củng cố:
- P1 — ViHOS: “ViHOS: Hate Speech Spans Detection for Vietnamese” — Hoang et al., EACL 2023.
    - P1 ghi nhận lack of context, allusion và boundary trong error analysis
- P2 — ViSoBERT: “ViSoBERT: A Pre-Trained Language Model for Vietnamese Social Media Text Processing” và P3 — VLUE/CafeBERT: “VLUE: A New Benchmark and Multi-task Knowledge Transfer Learning for Vietnamese Natural Language Understanding” đều dùng contextual pretrained representations, nhưng M5.2 đã chốt rằng chúng không isolate trực tiếp contextual-insult cases.
- Vì vậy direct literature support chính của H1 vẫn là P1 — ViHOS. Handoff cũng đã có candidate H1 theo đúng hướng này.

**Cơ chế đang nghi ngờ**
- Một representation chỉ biết: token = "nó", không biết token đó đang được dùng là trung tính, thân mật, xúc phạm hay hạ thấp. Contextual representation có khả năng tạo representation theo `token + surrounding words + sentence context`

**Hypothesis chính thức:**
- Mô hình gnas nhãn chuỗi có biểu diễn ngữ cảnh sẽ xử lý ViHOS tốt hơn ở một baseline phụ thuộc chủ yếu vào lexical/surface information, và lợi cish này đặc biệt nên xuất hiện ở các trường hợp context-dependent và hard-clean false-positive controls

**Điều gì làm H1 yếu đi hoặc bị bác ?**
- Contextual model không tốt hơn baseline hoặc tốt hơn chỉ ở direct insults như Ngu, hổn láo trong khi context-dependent/hard-clean vẫn không cải thiện thì ta không được nói improvement đến từ hiểu context

# 2. H2 - Social-Media Domain Hypothesis

**Evidence**
- S01: ko
- S02: punctuation ----CÚT
- S07: mắc dạy
- S09: ko, cmnr
- S12: báo đời
- S19: nông cản, auto sủa
- S22: lều báo
- S29: noisy/ambiguous spelling
- S44: ngáo
- S45: tào lao
- S48: Ko

**Literature quan trọng nhất**
- P2 — ViSoBERT: “ViSoBERT: A Pre-Trained Language Model for Vietnamese Social Media Text Processing” — Nguyen et al., EMNLP 2023.
    - Đây là paper có connection trực tiếp nhất

**Cơ chế giả định**

Nếu một pretrained model từng nhìn thấy nhiều dạng: ko, k, khum, cmnr, slang, emoji, misspelling trong pretraining, representation của nó có thể robust hơn với các dạng social-media text

**Hypothesis chính thức**

- H2 - Một mô hình được pretrain chuyên biệt trên văn bản mạng xã hội tiếng Việt có thể xử lý tốt hơn các trường hợp slang, teencode, abbreviation và noisy writing so với một mô hình tiếng Việt/general-domain không chuyên social media

# 3. H3 - Boundary Difficulty Hypothesis

H3 khác 2 hypothesis trên, nó không nhất thiết yêu cầu một model mới, nó là diagnostic hypothesis. Evidence cho thấy các câu hỏi đôi khi không phải có offensive không, mà là span phải batwsd dầu và kết thúc chính xác ở 

P1 cũng ghi nhận wrong span boundary là một loại lỗi quan trọng

**Hypothesis chính thức**
- H4 — Một phần đáng kể lỗi của các hệ thống ViHOS sẽ là lỗi span extent/boundary: model nhận diện được vùng offensive nhưng dự đoán thiếu hoặc thừa token, thay vì hoàn toàn không phát hiện offensive content.
