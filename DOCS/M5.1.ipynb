# Chốt các observation từ manual 50

# 1. Nghĩa offensive nhiều trường hợp không nằm sẵn trong tưng từ; nó được quyết định bởi ngữ cảnh

Đây là observation mạnh nhất từ manual 50. Trong workbook FINAL, 26/50 mẫu được gắn `Context-dependent`. Ngoài ra có 11 mẫu liên quan `Inappropriate pronount/address`. 

Evidence cụ thể:
| Sample | Gold / phần đáng chú ý | Ta quan sát được gì? |
|---|---|---|
| **S05** | `có vấn đề về quang học` | Nếu chỉ nhìn cụm này thì có thể là một nhận xét bình thường. Nhưng câu sau *“Con này là con mực nhé”* biến nó thành lời mỉa mai về khả năng nhìn nhận. |
| **S13** | `người khuyết tật` | Đây là cụm hoàn toàn trung tính trong nhiều ngữ cảnh. Trong câu này nó được dùng mỉa mai để hạ thấp người khác. |
| **S18** | `Thằng \| mày \| điên` | `điên` khá rõ, nhưng `Thằng`, `mày` không phải lúc nào cũng offensive. Sắc thái phụ thuộc quan hệ giao tiếp và cách dùng. |
| **S20** | `Con \| xấu \| xấu \| khẩu nghiệp \| nó` | `Con`, `nó` là từ tham chiếu thông thường nhưng trong toàn câu chúng góp phần tạo cách gọi thiếu tôn trọng. |
| **S32** | nhiều `Nó`, `Con` + `coi trời bằng vung` | Đây là case rất rõ: model không thể xử lý từng token độc lập. Thành ngữ và chuỗi tham chiếu phải được hiểu trong cả phát ngôn. |
| **S44** | `bọn \| ngáo` | `bọn` có thể trung tính trong câu khác; ở đây kết hợp với cách nói chê bai tạo sắc thái hạ thấp. |
| **S45** | `Mày \| tào lao \| nó` | `Mày`, `nó` tiếp tục cho thấy pronoun/address-term không có nhãn offensive cố định ngoài context. |

**Từ các mẫu đó rút ra được gì ?**
- Không phải đại từ `mày`, `nó`, `con`. `thằng` là hate words, cũng không phải từ `người khuyết tật` là offensive. Observation đúng là:
    - Trong manual 50, cùng một lexical item có thể trung tính hoặc offensive tùy cách nó được dung trong phát ngôn. Vì vậy việc nhận diện span không thể chỉ dựa trên một ánh xạ cố định `word -> offensive/non-offensive`

Đây là property có khả năng phân biệt trực tiếp giữa hai họ giải pháp sau này:
- Hệ thống chủ yếu dựa vào surface/lexical signal
- Hệ thống biểu diễn token trong context

# 2. ViHOS có các offensive spans mang nghĩa ẩn dụ, mỉa mai

Trong manual 50, 14/50 mẫu được audit với tag `Figurative/metaphor`. Đây không phải vài case rất giống nhau, chúng có nhiều dạng

Evidence cụ thể:

| Sample | Gold / biểu thức | Vấn đề |
|---|---|---|
| **S09** | `não bị vi rút ăn hết cmnr` | Không phải mô tả y khoa literal. Cả hình ảnh “virus ăn não” là cách hạ thấp trí tuệ. |
| **S15** | `thầy bar ko có não vẫn sống như thường` | “Không có não” được dùng figuratively để công kích khả năng suy nghĩ. |
| **S26** | `Trước khi làm gì cũng nên lắp não vào` / `Để não thành hoá thạch` | Hai span đều là hình ảnh ẩn dụ về trí tuệ, không phải profanity cố định. |
| **S27** | `miệng mở ra thúi như cái ống cống` | Offensive meaning nằm ở cả phép so sánh, không nằm riêng ở `miệng`, `thúi` hay `ống cống`. |
| **S35** | `con chó điên` / `sủa` / `cắn` | Nếu chủ thể thật sự là chó, `sủa`, `cắn` hoàn toàn bình thường. Chúng offensive vì đang được ánh xạ từ động vật sang người. |
| **S40** | gold `miệng rộng` | Toàn comment là phép nói phóng đại *“cái miệng rộng quá đẻ con ra còn lọt”*. Đây đồng thời là metaphor và vấn đề boundary. |
| **S42** | `Óc nó được nhồi đậu phụ thối` | Không có một bad word điển hình; cả hình ảnh mới tạo ra insult. |
| **S47** | `Mấy thằng chó` / `sủa bậy` | Tiếp tục dùng trường nghĩa động vật để hạ thấp người. |
| **S49** | `sủa` / `đi học tiếng người` | Offensive meaning xuất hiện từ đối lập “động vật ↔ người”, chứ không phải vì từ `sủa` intrinsically toxic. |

**Điểm quan trọng**

S09, S26, S27, S35, S42, S49 cùng cho ta một pattern: Một chuỗi các tư thông thường có thể trở thành insult khi chúng kết hợp thành một hình ảnh figurative

Ví dụ:
- `não + virus + ăn`: từng từ không phải bad words
- `miệng + ống cống`: cũng vậy
- `sủa`: là động từ bình thường khi subject là chó

Do đó observation hợp lệ là:
- Một phân các case trong manual 50 mã hóa tính offensive ở cấp cụm/cấu trúc figurative, thay vì nằm ở một lexical trigger đơn lẻ

Ta chưa kết luận được:
- ViHOS chủ yếu là metaphor, không có evidence đại diện để nói vậy. Cũng chưa được nói là Transformer sẽ giải quyết metaphor, đó là hypothesis sau này, không phải observation

# 3. Dữ liệu mang rõ đặc trưng noisy Vietnamese social media

Trong manual workbook:
- `Slang`: 10 mẫu
- `Slang/teencode`: 1 mẫu
- `Noisy spelling`: 3 mẫu
- Còn có punctuation nhiễu như S02

Evidence cụ thể:

| Sample | Dạng noise | Ví dụ |
|---|---|---|
| **S01** | abbreviation | `ko` |
| **S02** | punctuation distortion | `tiền----CÚT` |
| **S07** | spelling error + slang | `mắc dạy` thay vì dạng chuẩn `mất dạy`; `đớp` dùng nghĩa bóng |
| **S09** | teencode / abbreviation | `ko`, `cmnr` |
| **S12** | slang | `báo đời` |
| **S14** | slang / social meaning | `con bóng` |
| **S17** | colloquial/slang | `sân si`, `má`, `cút xéo` |
| **S19** | misspelling + web slang | `nông cản`, `auto sủa` |
| **S20** | slang | `khẩu nghiệp` |
| **S22** | wordplay/slang | `lều báo` |
| **S29** | noisy/ambiguous spelling | `Gia` có thể là tên hoặc lỗi viết của `già` |
| **S42** | slang-style figurative expression | `óc ... đậu phụ thối` |
| **S44** | slang | `ngáo` |
| **S45** | slang | `tào lao` |
| **S48** | abbreviation | `Ko` |

**S07 la một case rất đáng giữ**
- Có hai vấn đề khác nhau cùng lúc:
    - `mắc dạy` là noisy spelling
    - `đớp` là từ phải hiểu theo nghĩa bóng trong context
- Tức là noise không chỉ là vấn đề sửa chính tả, nó còn tương tác với semantics

**S09**
- Ở đây cùng lúc có metaphor, sarcasm, context, `cmnr` dạng viết tắt/teencode
- Đây là loại case mà ta sẽ rất cần khi M5.2 hỏi: Có lý do gì tư data của chính ta để nghiên cứu domain-specific social-media representation hay không

Observation hiện tại là:
- Manual 50 chứa nhiều dạng lexical/orthographic variation đặc trưng của social media: abbreviation, slang, teencode, spelling noise và punctuation distortion

# 5. Boundary là một difficulty độc lập, không đống nhất với việc "biết câu này offensive hay không"

S02:
- Comment: `Thằng điên văn vở câu like rẻ tiền----CÚT` => Gold: `Thằng điên | tiền----CÚT`. Ở đây CÚT rất rõ là offensive, nhưng `tiền----CÚT` làm boundary bị nhiễu bởi từ đứng trước và dấu gạch. Điều khó không phải phát hiện insult, điều khó là xác định chính xác extent

S04:
- Comment: `... chửi thẳng luôn: AD NGU` => Gold: `chửi | thẳng | NGU`. NGU rất rõ, nhưng chửi và đặc biệt thẳng chủ yếu thuộc cụm chửi thẳng mô tả cách nói

S24:
- Comment: `Khẩu nghiệp quá. Con cháu nghe xấu hổ chết` => Gold: `Khẩu nghiệp | chết`. Khấu nghiệp có cơ sở rõ, nhưng chết trong xấu hổ chết bản thân không phải insult rõ

S40:
- Comment: `cái miệng rộng qua đẻ con ra còn lọt` => Gold: `miệng rộng`. 

Observation:
- Với một số manual examples, khó khăn không nằm ở việc xác định có offensive content hay không, mà nằm ở việc xác định chính xác token nào phải thuộc span và span phải dài đến đâu

# 6. Clean sample cho thấy negative/suspicious vocabulary không đủ để quyết định offensive

S25:
- Comment: `súng bắn không trúng, dao chém không lủng`: Một keyword system có thể thấy súng, bắn, dao, chém và coi đây là toxic. Nhưng toàn câu là trích dẫn mang tính hài hước, đây là một hard negative cực tốt

S37:
- `không hay là phải`: Đây là đánh giá âm nhạc, nó tiêu cực nhưng không đánh giá con người, do đó criticism khác offensive attack

S39: 
- `... vui vẻ nhé và đừng quên xóa kênh`: Nếu cắt riêng đừng quên xóa kênh nghe khá hostile. Nhưng whole comment là lời chúc + joking trong community context

S48:
- `Em tưởng anh biến mất luôn cho em mừng`: Nếu chỉ đọc fragment này, rất dễ đánh false positive. Nhưng chính comment còn giải thích: `nội dung mang tính giải trí. Ko có ý anti hay ghét ai`. Do đó toàn ngữ cảnh quyết định nó là clean


Observation:
- Manual 50 có các clean examples chứa negative sentiment, criticism, violent vocabulary hoặc hostile-looking phrases nhưng không phải offensive. Vì vậy surface negativity là một shortcut không đáng tin cậy
